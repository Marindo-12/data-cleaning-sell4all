j'ai utiliser **claude** ici pour avoir les fonctions prédifinis **mean()** pour le moyenne et **median()** pour le mediane.

Pour cette question, j'ai pensé en premier lieu a utiliser une boucle, mais il donne un resultat different, apres recherche je trouve la fonction **Goupby** qui donne un resultat propre en minimum code.

```python
import matplotlib.pyplot as plt

depenses = iris.groupby('Country')['Customer spendings'].sum()

plt.figure(figsize=(10, 6))
depenses.plot(kind='bar', color='red')

plt.title('Depenses des clients par pays')
plt.xlabel('Pays')
plt.ylabel('Total des depenses')
plt.tight_layout()

plt.show()
```

ici on a afficher les graphiques qui montrent la somme des depenses, on peut aussi traiter a partir de moyenne ou mediane en changeant **sum()** par **mean()**. pour la fonction **tight_layout()**, il sert ajuster automatiquement l'espacement.

```python
#lignes_net_10 = iris['Customer spendings'] < 10
#print(lignes_net_10)

#lines_i10 = iris[iris['Customer spendings'] < 10]
#print(f'Nombre de lignes ayant dep < 10 :\n {lines_i10}')

iris = iris[iris['Customer spendings'] >= 10]

print(f'Nombre de lignes apres filtrages: {iris.shape[0]}')
```

L'idee c'est qu'on filter pour que le dataframe iris contient seulement les lignes avec depenses >= 10, avant , les lignes 505, maintenant 502, ce qui indique que 3 lignes en une depense < à 10.

```python
print(f"Nombre de lignes avant suppression des doublons: {iris.shape[0]}")

#iris = iris - iris.duplicated().sum()
iris = iris.drop_duplicates()

print(f"Nombre de lignes apres suppression des doublons: {iris.shape[0]}")
```
