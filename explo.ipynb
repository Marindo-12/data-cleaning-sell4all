```python
import pandas as pd

url = 'dataset-sell4all.csv'

colonnes = ['Name','Phone Number','Email','Address','Country','Postal code','Last date of connection','Last time of connection','Age','Gender','Customer spendings']

iris = pd.read_csv(url, header=0, names=colonnes)

print(iris.head())
```

```python
print('***************************')
print(f'-> Le nombre de lignes est: {iris.shape[0]}')
print('***************************')
print('-> Les colonnes de fichiers: ')
for i in iris.columns:
  print(f'- {i}')
print('***************************')
print('-> Les types de donnees: ')
for c,tc in iris.dtypes.items():
  print(f'- {c} : {tc}')
print('***************************')
```

## Résume Technique
**Nombre de lignes :** Le fichier contient 505 lignes correspondant chacune à un client de Sell4All.

**Colonnes diponibles :** Le fichier contient 11 colonnes.

**Types de données présents :**
- **str** : pour les colonnes textuelles comme Name...
- **int64** : pour les colonnes numériques int comme age...

Concernant le resultat de code **iris.info()**, il affiche un colonne de **Non-Null Count** dont on voie que tous les colonnes de dataset sont 505 non-null , ce qui signifie que tous les clients ont tous les donnees remplis.
