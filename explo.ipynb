```python
import pandas as pd

url = 'dataset-sell4all.csv'

colonnes = ['Name','Phone Number','Email','Address','Country','Postal code','Last date of connection','Last time of connection','Age','Gender','Customer spendings']

iris = pd.read_csv(url, header=0, names=colonnes)

print(iris.head())
```

```python
print('***************************')
print(f'-> Le nombre de lignes est: {iris.shape[0]}')
print('***************************')
print('-> Les colonnes de fichiers: ')
for i in iris.columns:
  print(f'- {i}')
print('***************************')
print('-> Les types de donnees: ')
for c,tc in iris.dtypes.items():
  print(f'- {c} : {tc}')
print('***************************')
```

```python
moyenne_age = iris['Age'].mean()
mediane_age = iris['Age'].median()

moyenne_cs = iris['Customer spendings'].mean()
mediane_cs = iris['Customer spendings'].median()

print('++++++++++++++++++++++ Age +++++++++++++++++++++++++')
print(f'-> Moyenne de colonne age: {round(moyenne_age,2)}')
print(f'-> Mediane de colonne age: {mediane_age}')
print('+++++++++++++++ Customer Spendings +++++++++++++++++')
print(f'-> Moyenne de colonne Customer spendings: {round(moyenne_age,2)}')
print(f'-> Mediane de colonne Customer spendings: {mediane_age}')
print('++++++++++++++++++++++++++++++++++++++++++++++++++++')
```

**Age et Customer spendings ont les memes valeurs de mediane et moyenne**

j'ai utiliser **claude** ici pour avoir les fonctions prédifinis **mean()** pour le moyenne et **median()** pour le mediane.

```python
med_age_pay = iris.groupby('Country')['Age'].median()

print('~~~~~~~~~~ Médiane d\'age pour chaque pays ~~~~~~~~~~~')
print(med_age_pay)
```

Pour cette question, j'ai pensé en premier lieu a utiliser une boucle, mais il donne un resultat different, apres recherche je trouve la fonction **Goupby** qui donne un resultat propre en minimum code.

```python
import matplotlib.pyplot as plt

depenses = iris.groupby('Country')['Customer spendings'].sum()

plt.figure(figsize=(10, 6))
depenses.plot(kind='bar', color='skyblue')

plt.title('Depenses des clients par pays')
plt.xlabel('Pays')
plt.ylabel('Total des depenses')
plt.tight_layout()

plt.show()
```
